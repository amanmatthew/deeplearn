```python
#Import the libraries 

import tensorflow as tf 
import numpy as np 
import pandas as pd 
import time 
from tensorflow.keras.datasets import imdb 
from tensorflow.keras import *
from tensorflow.keras.layers import SimpleRNN, LSTM, GRU, Input, Embedding, Dense
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
#Train the dataset 

(xtrain, ytrain), (xtest, ytest) = imdb.load_data(num_words=10000)
xtrain, ytrain = xtrain[:10000], ytrain[:10000]
xtest, ytest = xtest[:2000], ytest[:2000]

xtrain = pad_sequences(xtrain, maxlen=200)
xtest = pad_sequences(xtest,maxlen=200)
```

```python
#Build the model 

def build_model(layer_name):
    if layer_name == "SimpleRNN":
        rnn_layer = SimpleRNN(32)
    elif layer_name == "LSTM":
        rnn_layer = LSTM(32)
    else:
        rnn_layer = GRU(32)

    model = Sequential([
        Input(shape=(2000, )),
        Embedding(10000,2),
        rnn_layer,
        Dense(1, activation="sigmoid")
    ])

    model.compile(
        optimizer="adam",
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )
    return model
```

```python
# Train and evaluate the model

results = {}

for name in ["SimpleRNN", "LSTM", "GRU"]:
    model = build_model(name)

    start = time.time()
    model.fit(xtrain, ytrain, epochs=5, validation_split=0.2, verbose=0)
    test_time = time.time() - start

    loss, accuracy = model.evaluate(xtest, ytest, verbose=0)

    results[name] = {
        "accuracy" : round(accuracy*100, 2),
        "loss" : round(loss, 4),
        "time" : round(test_time, 2)
    }

# Print the RNN LSTM GRU Comparisons

line = "=" * 60

print(line)
print("         RNN LSTM AND GRU COMPARISONS")
print(line)
print(f"{'MODEL':<14}{'TEST ACCURACY':<20}{'TEST LOSS':<14}{'TRAINING TIME'}")
print("-" * 60)

for name, r in results.items():
    print(f"{name:<14}{r['accuracy']:<20}{r['loss']:<14}{r['time']}")

print(line)
```

```python
# Print the Final Evaluation

best_acc = max(results, key=lambda m:results[m]['accuracy'])
best_loss = min(results, key=lambda m:results[m]['loss'])

print("       FINAL ANALYSIS")
print(line)
print("BEST MODEL ACCURACY : ",best_acc)
print("BEST MODEL LOSS :",best_loss)
print(line)
```
