```python
# Import the libraries 

import tensorflow as tf 
from tensorflow.keras.datasets import imdb 
from tensorflow.keras import *
from tensorflow.keras.layers import Input, Embedding, SimpleRNN, Dense
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
# Load the imdb dataset 

(xtrain, ytrain), (xtest, ytest) = imdb.load_data(num_words=10000)
xtrain, ytrain = xtrain[:10000], ytrain[:10000]
xtest, ytest = xtest[:2000], ytest[:2000]

xtrain = pad_sequences(xtrain, maxlen=200)
xtest = pad_sequences(xtest, maxlen=200)
```

```python
# Train the model 

model = Sequential ([
    Input(shape=(2000,)),
    Embedding(10000,32),
    SimpleRNN(32),
    Dense(1, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.fit(xtrain, ytrain, epochs=5, batch_size=64, verbose=0)
loss, accuracy = model.evaluate(xtest, ytest, verbose=0)
scores = model.predict(xtest, verbose=0)
```

```python
print("Accuracy : ",accuracy*100)
print("Loss : ",loss)

for i in range(5):
    if ytest[i] == 1:
        actual = "Positive"
    else:
        actual = "Negative"

    if scores[i][0] >= 0.5:
        predicted = "Positive"
    else:
        predicted = "Negative"

    print("-" * 60)
    print("FINAL RESULTS")
    print("Actual Sentiment : ",actual)
    print("Predicted Sentiment : ",predicted)
    print("-" * 60)
```
